```python
#Import the packages

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from CBFV import composition

from sklearn.linear_model import LinearRegression

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import normalize
```

```python
data=pd.read_csv("data/lec05-HW_formation_energy.csv")
data.head()
```

```python
len(data) #that's a lot of data
```

```python
data.isna().sum()
```

```python
#There are two missing formulas, out of a total of 55K!, so let's just drop them

data=data.dropna(axis=0,how='any')
len(data)
```

```python
#There are a lot of un-necessary columns, let's get rid of some of them for convenience
data=data.drop(['Unnamed: 0','optb88vdw_bandgap','func','spg_symbol','spg_number','jid'],axis=1)
data.head()
```

```python
#Remember: if we want to use CBFV to featurize the data, we need to have our columns as 'formula' and 'target'
data=data.rename(columns={'formation_energy_peratom':'target'})
data.head()
```

```python
print(f'shape of data {data.shape}')
```

```python
X,y,formulae,skipped=composition.generate_features(data,elem_prop='oliynyk')  
#You can try different elem_prop arguments such as -mat2vec, jarvis etc.
```

```python
X.head()
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.2)
len(X_train)
```

```python
scaler=StandardScaler()  #Instantiate a scaler

X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
model=LinearRegression()  # Create an instance of Linear Regression
result=model.fit(X_train,y_train)
```

```python
r2=model.score(X_train,y_train)
print(r2)
```

```python
r2_test=model.score(X_test,y_test)
print(r2_test)
```

We said that the problem was the length of the dataset in comparison to the number of features
In-class problem: train data - 46 while no. of features - 132 (magpie featurizer)

The dataset $(n)$ is too small $<<$ compared to features $(p)$  
In that case the equations are underdetermined - you can always find a solution that fits best to the training data.
So we have to significantly drop featurizers so that $n>p$ for this to work.

Whereas, 

In this problem : train data - 44571, while no.of features - 264 (oliynik featurizer)
So, $n>>p$. 
Therefore we don't have the same problem - the fitting works
