```python
import pandas as pd
import numpy as np
from CBFV import composition

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv("data/lec05_Thermal_Cond.csv")
data.head()
```

```python
data=data.rename(columns={"Compound":"formula","K (W/mK)":"target"})
data.head()
```

```python
X,y,formulae,skipped = composition.generate_features(data,elem_prop='magpie')
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=23)
```

```python
linear_model = LinearRegression()
linear_model.fit(X_train,y_train)
```

```python
r2_train=linear_model.score(X_train,y_train)
print(r2_train)
```

```python
from sklearn.metrics import r2_score
```

```python
y_train_pred = linear_model.predict(X_train)
r2_train_value=r2_score(y_train,y_train_pred)
print(r2_train_value)
```

```python
y_test_pred =linear_model.predict(X_test)
r2_test_value=r2_score(y_test,y_test_pred)
print(r2_test_value)
```

**Learning Point!**
Lots of features (descriptors) does not always make good models.


*What's happening here*

We have ~48 training samples and 132 features! There are more unknowns than equations!
e.g. x+y=5 : you can find an infinite number of values for x & y that satisfy this equation. 
That means, you can find a solution that runs through every datapoint exactly!
That's why r2_train=1 --> overfit, but when you look at the test data r2 verlow/negative : terrible!

Lessons!
- if r2=1 or very close to 1 --be very suspicious: check what's wrong!
- if r2_train is high, and r2_test is low --> overfitting to training data
- if r2_test is high and r2_train is significantly low --> that's very strange but could be due to a sampling issue (should try cross-validation instead of simple train-test split)
- if r2_train and test are similar --> very good!
- more features does not always mean a better model - it depends on the data volume / problem complexity and model complexity


# What are we trying to do?

Predict Thermal conductivity using known features.
But what parameters do K depend on? - in the microscopic regime (not bulk properties) - things like
- atomic/crystal structure (phonons)
- free electron density
- atomic mass
- defects and impurities
- porosity
  are thought to play a role - empirically!

Often we use ML to investigate unknown relationships in physics/chem/mmaterials.
If the relationships are unknown - how do we know which features we need for a good prediction??
**This is a big problem in featurization in classical machine learning**
Which can be overcome/ minimized through Deep learning.

Another important point-
*Defects in materials*: Which plays a major role in many aspects - from property prediction (like this example) to material failure preedictions - but how can we incorporate this in the model??

## Summary of learnings

- We need to investigate how 'meaningful' the task is in terms of science.
  - Does the physics make sense? What are the 'known' descriptors, how accurately can we represent them, what are the limitations, what are the 'unknown' possibilities, can we test their predictability etc.
- Think about the 'meaning' of the problem
- Featurizing the dataset is always preferred to extract chemical information from the formulas.
- Collect data - size of dataset should be much much bigger than the number of features we wish to use (the guideline is at least 10 datapoints per feature - so if you have *n* features, you need *10n* datapoints (minimum) - but this depends on a few other factors  
   - Use dimensionality reduction techniques to find meaningful features / Or use deep learning
- Look out for *perfect fits* a model that predicts the target almost perfectly is too good to be true - there must be something wrong!
   - Try different metrics
   - Try metrics on both train & test splits (they should be similar)
   - Use cross-validation (later)
   - Use models like Ridge regression instead of linear regression (later)
