# Closing thoughts: learning to adapt 💪

We have reached the end of the course. Before you go, here is one idea that may matter more than any single method you have learned here.

## The ground will keep moving

Technology will change, and it will change quickly during your working life. Developments in AI will change the technical side of things first. That, in turn, will change how businesses operate, how universities are organised, and how medicine and research are carried out. It will shape the global economy, how people interact, and even geopolitics.

Change will be a normal part of life, it already is. Your success will not depend on one fixed set of knowledge and skills. It will depend on how well you can **learn** and **adapt**.

## Build on strong foundations 🏗️ 

So how do you learn and adapt? 🧠 Ground your understanding in the basic sciences: mathematics, physics and chemistry.

🧪 🔭🔧 ➕➖✖️➗🧑‍🔬⚛️

Whatever else changes, the fundamentals stay more or less the same. There are occasional overhauls, but they are rare and even if there is an overhaul, our current understanding will still stand. Think of a building. If the foundation is solid, it does not matter if the walls have to be knocked down and rebuilt. You can always rebuild them.

You have already seen this in the course. The newest-looking ideas rested on old foundations:

- A single **neuron** is linear or logistic regression with an activation function.
- **Training** a neural network is minimising a loss, just as least squares did for linear regression.
- **Backpropagation** is the chain rule from calculus, applied many times.
- **Gradient descent** and **Adam** are ways of walking downhill on a surface you can describe with derivatives.
- **Machine Learning** itself is based on ideas of statistics and linear algebra.
- **Python** is still based on the conceptual framework of object-oriented-programming.
  
Frameworks and model names will come and go. These foundations will not.🏛️

## Science as a way of thinking 💭

Carl Sagan made the point that science is more than a collection of facts: it is a way of thinking.[^sagan]

If you know how to *think* 🤔- to ask how something works ❓, to test it, and to change your mind when the evidence says so - then you know how to learn. And if you know how to learn, you can accept change and adapt to it.

[^sagan]: Paraphrased from C. Sagan, *The Demon-Haunted World: Science as a Candle in the Dark* (Random House, 1995).

## What stays, what goes

Some parts of today's AI will stay with us and may change everyday life. Some will fade away. New trends will appear that nobody can predict yet.

But as long as you understand the inner workings - how the maths works, how the code works, and how the physics or chemistry works (or the biology, if that is your field) - you will be able to make sense of whatever comes next.

:::{tip} Habits worth keeping
- When you meet a new method, ask: *which fundamentals is this built on?*
- Try to rebuild a small version yourself, based on what you know and then add to it.
- Check new claims against what you know from physics, chemistry and the data.
:::

Good luck 👍- and keep learning!.

Always stay curious!
