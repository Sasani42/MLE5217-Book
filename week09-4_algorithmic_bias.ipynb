# A serious note: algorithmic bias 	😐

A model is only as fair and as accurate as the data it learns from. **Algorithmic bias** is when a system makes *systematic*, repeatable errors that treat one group unfairly - usually because the training data wasn't representative, or because the model latched onto a misleading shortcut.

Some widely reported examples:

- Google Photos, 2015: Google’s new Photos app labelled photos of Black people as “gorillas”. It came to light when software engineer Jacky Alciné posted about photos of himself and a friend. Google apologised and blamed too few photos of Black people in its training data. Its fix was to stop the app labelling anything as a gorilla - which doesn't fix the root problem, and a 2023 New York Times check found that Google Photos still blocked gorilla, monkey and ape labels. (PetaPixel, summarising the NYT report; TheGrio, 2015)

 - Gender Shades, 2018: Buolamwini and Gebru tested three commercial gender-classification systems on a new dataset balanced by gender and skin type, using the Fitzpatrick scale. Darker-skinned women had error rates of up to 34.7%, against at most 0.8% for lighter-skinned men. (PMLR proceedings)


- **Skin lesions.** 

<img src="images/lec09_skin.png" width="400">

Researchers fed a model with skin lesion images – some Were cancerous, others were not. The model was asked to classify malignant Vs benign.The idea was to develop a model which can simply look at a 
Lesion and flag if it’s cancerous or not.

However, what they didn't account for is  that images of some lesions included a **ruler** for scale - in the practical scenario, when a medical professional examined the patient he/she can make an experienced judgement on whether the lesion is cancerous or not. If the doctor suspected this could be cancerous, he/she would place a ruler next to it when taking the image, so that the growth can be quantified in a future visit. However, if the doctor was sufficiently confident that the lesion is benign, he/she would not place the ruler.

So when these images were fed, the model quietly learned "ruler → malignant" instead of learning the lesion itself (Narla et al., 2018).
So it was leaning on human classification rather than looking at the lesions features itself.

- **Husky vs. wolf.** 
A similar incident took place with a research involving building a model to classify the husky breed of dogs and wolves. 
 
<img src="images/lec09_husky.png" width="400">


The model did well to separate Huskies and wolves but when researchers closely inspected the incorrect classifications to better understand the model – they suspected something, and then used tools to look at what the model was basing its decisions on.

It turned out that apart from looking at the features of the animal, the model was also looking at the background of the image. If there was snow in the background the model classified it as a wolf! Simply because wolves are most likely to be found in snowy environments.

(Ribeiro et al., 2016). It wasn't recognising the animal at all.

Bias can also **feed on itself**. Recommendation systems can trap people in bubbles; predictive-policing tools (like PredPol in USA) kept sending patrols to the same neighbourhoods because past arrest data pointed there - which then generated more of the same data.


## What does algorithmic bias mean for materials science?

Materials ML rarely harms a demographic group directly the way face recognition can. The bias usually enters through what data exists, how models are evaluated and what we choose to optimise. The result is skewed discoveries and overconfident claims.

**Human choices shape what gets made and tested** 
Chemists tend to reuse familiar reagents and a narrow range of conditions, so the recorded data cover a small, popular corner of the possible space. Jia et al. (2019, Nature) showed this for inorganic synthesis. Models trained on randomly generated experiments predicted reaction outcomes better than models trained on human-chosen experiments, and were more useful for finding new regions to explore. (Haverford record; authors’ summary)

**Failed experiments go unreported**
Journals publish successes, so a model trained on literature data sees mostly positives and can’t learn where things fail. Raccuglia et al. (2016, Nature) trained a model on archived “dark” reactions, failed syntheses from lab notebooks. It predicted successful crystallisation of new templated vanadium selenites 89% of the time, outperforming human intuition. (Haverford record; Phys.org)

**Simulation methods carry systematic errors**
Large computed databases are typically built with one DFT approach. A model trained on them learns that method’s systematic errors as if they were real properties. The classic example is the PBE functional underestimating band gaps. Mixing data from different functionals, or from computation and experiment, without accounting for this adds a hidden offset.

**Random cross-validation overstates discovery power** 
Materials datasets contain families of near-duplicates. Random k-fold CV puts close relatives in both the training and test sets, so the scores reflect interpolation, not discovery. Meredig et al. (2018) proposed leave-one-cluster-out CV for this. For predicting superconductor critical temperature, a random forest’s correlation fell from R = 0.87 under ordinary CV to a median of 0.30 when whole clusters were held out. YBCO went from a predicted high-Tc material to a below-average one once no cuprates were in training. (RSC, open access)

We have been talking about this during this course, and you would have dealt with this in your assignments. 


**Instruments and labs leave fingerprints**
Images, spectra or mechanical data from one microscope, beamline or lab’s processing route can teach a model the instrument’s signature instead of the material’s. It then fails on another lab’s data. Human labels of microstructures (phase, grain boundaries, defects) also carry annotator judgement.

**Choosing what to optimise is a value judgement**
A model that optimises strength or efficiency alone will steer toward whatever wins on that metric. That can mean expensive, toxic, supply-critical or hard-to-recycle elements. Leaving cost, sustainability or safety out of the objective is a design choice, and it is where human impact enters materials ML most directly.

**Model-guided experiments reinforce themselves**
In active learning, the model suggests the next experiments, and those results feed back into training. Without deliberate exploration, it keeps sampling where it is already confident, deepening the skew from the first point.


**Practical tips for minimizing algorithmic bias in materials science** 

- Check how well the dataset covers chemical space before trusting a model,
- report failed and negative results,
- use grouped or cluster-based splits for any discovery claim,
- record the instrument, lab and calculation method alongside each data point, 
- state the full objective, including cost and sustainability, when using ML to screen candidates.

> Use **representative, balanced** training data, check your metrics **for each group**, and always know **where your model is (and isn't) reliable.**
